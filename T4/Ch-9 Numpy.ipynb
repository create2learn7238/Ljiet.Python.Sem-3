```python
import numpy as np
arr1=np.array([[3,4,2],[1,7,3],[9,0,5]])
print("Sort")
print(np.sort(arr1)) 
print()
print("Sort Axis 0")
print(np.sort(arr1,axis=0))
print()
print("Sort Axis 1")  # aama axis = dimension-1 ane sort ans same j aavse 😏😣
print(np.sort(arr1,axis=1))
print()
print("Sort Axis None") 
print(np.sort(arr1,axis=None))
```

## Random
- [0,100) 0 included and 100 excluded

```python
from numpy import random as r
x=r.randint(100) # Number from 0 to 99
print(x)
x=r.randint(10 , 100,size=5) # array of length 5
print(x)
x=r.randint(10 , 100,size=(3,5)) # 3 array of length 5 
print(x)
```

```python
from numpy import random as r
x=r.rand() # 0 to 0.99
print(x)
x=r.rand(5) # array of len 5
print(x)
x=r.rand(5,3) # 5  array of len 3
print(x)
```

```python
from numpy import random as r
x=r.choice([1,2,3,4,5,6,7]) # any one
print(x)
x=r.choice([1,2,3,4,5,6,7],size=(2)) # any array of 2
print(x)
x=r.choice([1,2,3,4,5,6,7],size=(5,3)) # 7 any array of 2
print(x)
```
