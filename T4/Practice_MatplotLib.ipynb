```python
import matplotlib.pyplot as plt
import numpy as np
plt.figure(figsize=[15,10])
x=np.random.randint(1,10,size=5)
y=np.random.randint(1,10,size=5)
#================
#1st Line Graph
#================
plt.subplot(2,3,1)
plt.plot(x,y,marker="o",ls=":",color="r",ms=10,mec="b",lw=2)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("1st")
plt.grid(c="green")
#================
#2st Bar Graph
#================
plt.subplot(2,3,2)
plt.bar(x,y,ls=":",color="b",lw=4)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("2nd")
plt.grid(c="orange")
#================
#3st stack
#================
plt.subplot(2,3,3)
plt.fill_between(x,y,ls=":",color="green")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("3nd")
plt.grid(c="red")
#================
#3st stack
#================
plt.subplot(2,3,4)
plt.scatter(x,y,color="orange")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("3nd")
plt.grid(c="yellow")
#================
#5st Pie
#================
plt.subplot(2,3,5)
plt.scatter(x,y,color="r")
l=['A','B','C','D','E']
plt.grid(c="pink")
plt.pie(y,labels=l)
#================
#6st Pie
#================

plt.show()
```

### WAP To Creat a bar plot og courses and a number of student enrol in their course usind the following dict with appropriate label and color of the bar should be green

```python
import matplotlib.pyplot as plt
data={"C":20,"C++":11,"JAVA":15,"Python":30,"Maths":45}
y=data.values()
x=data.keys()
plt.bar(x,y,color="green")
plt.grid(c="yellow")
plt.xlabel("Courses")
plt.ylabel("No of Students")
plt.show()
```

### Made same for python

```python
import matplotlib.pyplot as plt
data={"C":20,"C++":11,"JAVA":15,"Python":30,"Maths":45,"JS":10,"PHP":21,"R":14}
y=data.values()
x=data.keys()
plt.pie(y,labels=x,autopct="%1.1f%%",explode=[0.5,0,0,0,0,0,0,0.5])
plt.grid(c="yellow")
plt.show()
```
