## Multuple lines

```python
import matplotlib.pyplot as pit
import numpy as np
x=np.array([80,85,90,95,100,105,110,115,120,125]) 
y=np.array([240,250,260,270,280,290,300,310,320,330]) 
font1={'family':'serif','color':'blue','size':25}
font2={'family':'serif','color':'darkred','size':20}
pit.plot(x,y,"o--")
pit.xlabel("Average Pulse",font2)#
pit.ylabel("Calories burn",{'family':'veranda','color':'orange','size':20})
pit.title("Smart Watch",loc="left",fontdict=font1)
pit.grid(c="green")
pit.show()
```

## Subplot 

```python
# ROW , Column
import matplotlib.pyplot as pit
import numpy as np
x=np.array([0,1,2,3]) 
y=np.array([3,8,2,10]) 
pit.subplot(1,2,1)
font1={'family':'serif','color':'blue','size':20}
font2={'family':'serif','color':'darkred','size':10}
pit.plot(x,y,"o--")
pit.xlabel("Average ",font2)#
pit.ylabel("Calories ",{'family':'veranda','color':'orange','size':15})
pit.title("1st",loc="right",fontdict=font1)
pit.grid(c="green")
x1=np.array([1,4,2,6]) 
y1=np.array([3,9,1,8]) 
pit.subplot(1,2,2)
pit.plot(x1,y1)
pit.xlabel("Average ",font2)#
pit.ylabel("Calories ",{'color':'green','size':10})
pit.title("2nd ",loc="left",fontdict=font1)
pit.grid(c="green")
pit.suptitle("2 in 1")
pit.show()
```

## 6 in one

```python
# ROW , Column , number or tiles
import matplotlib.pyplot as pit
import numpy as np
pit.figure(figsize=[15,10])
x=np.array([0,1,2,3]) 
y=np.array([3,8,2,10]) 
pit.subplot(2,3,1)
pit.plot(x,y,marker="o",ls=":",color="r")
pit.xlabel("X 1")
pit.ylabel("Y 1")
pit.title("1st")

x=np.array([0,1,2,3]) 
y=np.array([3,8,2,10]) 
pit.subplot(2,3,2)
pit.plot(x,y,marker="v",ls="--",color="blue")
pit.xlabel("X 2")
pit.ylabel("Y 2")
pit.title("2nd")

x=np.array([0,1,2,3]) 
y=np.array([3,8,2,10]) 
pit.subplot(2,3,3)
pit.plot(x,y,marker="v",ls="-.",color="g")
pit.xlabel("X 3")
pit.ylabel("Y 3")
pit.title("3rd")

x=np.array([0,1,2,3]) 
y=np.array([3,8,2,10]) 
pit.subplot(2,3,4)
pit.plot(x,y,marker="o",ls=":",color="r")
pit.xlabel("X 4")
pit.ylabel("Y 4")
pit.title("4th")

x=np.array([0,1,2,3]) 
y=np.array([3,8,2,10]) 
pit.subplot(2,3,5)
pit.plot(x,y,marker="o",ls="-",color="darkred")
pit.xlabel("X 5")
pit.ylabel("Y 5")
pit.title("5th")

x=np.array([0,1,2,3]) 
y=np.array([3,8,2,10]) 
pit.subplot(2,3,6)
pit.plot(x,y,marker="v",ls=":",color="darkblue")
pit.xlabel("X 6")
pit.ylabel("Y 6")
pit.title("6th")

pit.suptitle("6 in one ")
pit.show()
```

### QUE. Area fill between two lines

```python
import matplotlib.pyplot as p
import numpy as np
x=np.arange(12)
y=np.array([5,9,6,6,10,7,6,4,4,5,6,4])
z=np.array([6,6,8,3,6,9,7,8,6,6,4,8])
p.plot(x,y,color='blue')
p.plot(x,z,color='red')
p.fill_between(x,y,z, where=(y>z),color="blue")
p.fill_between(x,y,z, where=(y<=z),color="red")
p.grid(color='green')
p.show()
```

```python
import matplotlib.pyplot as p
import numpy as np
x=np.arange(12)
y=np.array([5,9,6,6,10,7,6,4,4,5,6,4])
z=np.array([6,6,8,3,6,9,7,8,6,6,4,8])
p.plot(x,y,color='blue')
p.plot(x,z,color='red')
p.fill_between(x,y,z, where=(y>z),color="blue",label='Positive',alpha=0.7,interpolate=True)
p.fill_between(x,y,z, where=(y<=z),color="red",label="Negative",alpha=0.7,interpolate=True)
p.legend()
p.grid(color='green')
p.show()
```

### Stacked area chart

```python
import matplotlib.pyplot as p
import numpy as np
x=range(1,6)
y1=np.array([1,4,6,8,4])
y2=np.array([8,2,7,10,12])
y3=np.array([2,8,5,10,6])
p.stackplot(x,y1,y2,y3,labels=['A','B','C'],colors=['blue','green','yellow'],alpha=0.5)
p.legend(loc='upper left')
p.grid(color='green')
p.show()
```

### Pie chart

```python
import matplotlib.pyplot as p
import numpy as np
p.figure(figsize=[7,7])
y=np.array([35,25,25,15])
l=['A','B','C','D']
p.pie(y,colors=['red','green','black','yellow'],startangle=90,labels=l,explode=[0,0,0.15,0],radius=2)
p.legend(title="pieee")
p.show()
```

```python
import matplotlib.pyplot as p
import numpy as np
p.figure(figsize=[7,7])
y=np.array([30,20,50,40])
l=['A','B','C','D']
p.pie(y,colors=['red','green','orange','yellow'],labels=l,explode=[0,0,0.15,0],autopct="%1.1f%%")
p.legend(title="pieee")
p.show()
```

```python
import matplotlib.pyplot as p
import numpy as np
y=np.array([1]*24)
p.pie(y,colors=['white'],explode=[0.1]*24,wedgeprops={'edgecolor':'blue'})
p.show()
```
